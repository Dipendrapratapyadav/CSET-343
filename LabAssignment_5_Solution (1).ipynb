```python
df = df.drop_duplicates()

df["age"] = pd.to_numeric(df["age"], errors="coerce")

df["age"] = df["age"].fillna(df["age"].median())

print("Missing values:", df.isnull().sum().sum())
```

```python
df = df.drop_duplicates()

df["age"] = df["age"].fillna(df["age"].median())

print("Missing values:", df.isnull().sum().sum())

Q1 = df["age"].quantile(0.25)
Q3 = df["age"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Age outliers:", ((df["age"] < lower) | (df["age"] > upper)).sum())

df["age"] = df["age"].clip(lower, upper)
```

```python
import sklearn
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split

X = df.drop('class', axis=1)

le = LabelEncoder()
y = le.fit_transform(df['class'])

print("Encoding:", dict(zip(le.classes_, le.transform(le.classes_))))

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

print("Train:", X_train.shape)
print("Test:", X_test.shape)

scaler = StandardScaler()

X_train_s = scaler.fit_transform(X_train)
X_test_s = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "KNN": KNeighborsClassifier(n_neighbors=5),
    "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42),
    "SVM": SVC(probability=True),
    "Naive Bayes": GaussianNB()
}
results = []

for name, model in models.items():
    model.fit(X_train_s, y_train)
    y_pred = model.predict(X_test_s)

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, average="macro"),
        "Recall": recall_score(y_test, y_pred, average="macro"),
        "F1 Score": f1_score(y_test, y_pred, average="macro")
    })

results = pd.DataFrame(results)
results
```

```python
preds = {}

for name, model in models.items():
    model.fit(X_train_s, y_train)
    preds[name] = model.predict(X_test_s)
labels = [str(x) for x in le.classes_]
```

```python
for name in models:
    print("\n", name)
    print(classification_report(
        y_test,
        preds[name],
        target_names=labels,
        zero_division=0
    ))
```

```python
from sklearn.metrics import confusion_matrix,ConfusionMatrixDisplay
fig, axes = plt.subplots(2, 3, figsize=(15, 8))

for ax, name in zip(axes.ravel(), models):
    ConfusionMatrixDisplay(
        confusion_matrix(y_test, preds[name]),
        display_labels=le.classes_
    ).plot(ax=ax, colorbar=False)

    ax.set_title(name)

axes.ravel()[-1].axis('off')

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import roc_curve, auc
from sklearn.preprocessing import label_binarize

n_classes = len(le.classes_)
y_test_bin = label_binarize(y_test, classes=range(n_classes))
```

```python
probas = {}

for name, model in models.items():
    probas[name] = model.predict_proba(X_test_s)
```

```python
plt.figure(figsize=(7, 6))

for name in models:
    fpr, tpr, _ = roc_curve(
        y_test_bin.ravel(),
        probas[name].ravel()
    )

    plt.plot(
        fpr,
        tpr,
        label=f'{name} (AUC={auc(fpr, tpr):.3f})'
    )

plt.plot([0, 1], [0, 1], 'k--')

plt.xlabel('FPR')
plt.ylabel('TPR')
plt.title('Micro-average ROC')
plt.legend(loc='lower right')

plt.show()
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.pipeline import make_pipeline
skf = StratifiedKFold(5, shuffle=True, random_state=42)
cv = {n: cross_val_score(make_pipeline(StandardScaler(), m), X, y, cv=skf, scoring='f1_macro') for n, m in models.items()}
cv_df = pd.DataFrame({n: [s.mean(), s.std()] for n, s in cv.items()}, index=['CV F1 macro (mean)', 'std']).T.round(4)
cv_df
```
